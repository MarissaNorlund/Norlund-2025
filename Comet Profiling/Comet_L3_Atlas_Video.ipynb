```python
from IPython import get_ipython
from IPython.display import display


# Install necessary libraries
!pip install --upgrade prose
!pip install astropy


# Mount Google Drive
from google.colab import drive

drive.mount('/content/drive')


# Import necessary libraries
from prose import FitsManager, FITSImage, simulations, Sequence, blocks
import os
import matplotlib.pyplot as plt
import numpy as np
from prose.blocks.visualization import Video
from prose.core.source import ExtendedSource
from astropy.io import fits

# Set folder path for images
folder = "/content/drive/MyDrive/Comet/Comet/Comet L3 Atlas Jan.6.2022"

# Find all FITS files in the folder and subfolders
fits_files = []
for filename in os.listdir(folder):
    if filename.endswith(('.fits', '.fit')):
        filepath = os.path.join(folder, filename)
        fits_files.append(filepath)

# Print the found files and their count for verification
print("FITS files found:")
print(fits_files)
print(f"Total FITS files: {len(fits_files)}")

# Process the images to get the comet information first
# Create a FITSImage object using the first file for demonstration
image = FITSImage(fits_files[0])

# Create a Sequence for image processing
s = Sequence(
    [
        blocks.Trim((20, 20)),
        blocks.AutoSourceDetection(threshold=7),
        blocks.CentroidGaussian2D(cutout=31),
    ]
)

# Run the sequence on the image
s.run(image, show_progress=False)

# find the comet, extended object with largest area
area = [s.area if isinstance(s, ExtendedSource) else 0 for s in image.sources]
i = int(np.argmax(area))
print(f"Index of the comet in image.sources: {i}")


# Create Video block - this is moved to the end
video = Video(destination="video3.mp4", fps=10, width=500)

# Process images and generate video
sequence = Sequence([video])
sequence.run(fits_files, show_progress=True)
```
